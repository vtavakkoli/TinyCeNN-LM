## Evidence-based improvements over V5.3

Your V5.3 report: replacing layer 24 (64 channels, 3 recurrent updates) achieved **75.5% gold accuracy on 200 TEST decisions**, exactly the same as the compressed non-CeNN baseline (teacher **76.5%**). GPU inference was **0.989×** as fast as the compressed baseline, and the layer-24 cell had **247,233 parameters versus 233,472** in its compressed projection: this choice **increased** memory locally. V5.3 gold validation stalled at **83.33%** through step 140.

This notebook **does not repeat V5.3's memory mistake**: it searches replacement layers 8, 17, 20, 23, 26, never selects a replacement unless it actually reduces the individual projection's parameter count. It evaluates a cheaper **32-channel, 2-step liquid cell** across the candidate layers and tries **64-channel, 2-step** alternatives for the two strongest candidates. It checks the **whole 180-decision gold-validation set** before the final choice rather than relying on only 48 decisions.

### Architectural experiment

Replace one pretrained compressed FFN output `mlp.Wo` with a genuine lower-parameter `LiquidCeNNProjection`. The cell is a 1D token-lattice depthwise 3-tap neighborhood, input and recurrent transforms, and a bounded input-dependent relaxation update. All 28 pretrained ModernBERT layers and the other 20 selected SVD/pair-delta projections remain; the specific FFN output replaced by CeNN is one of those projections or an original full projection.

A complete compressed **baseline clone** is made via the original weights in `state_dict`, without recomputing expensive SVD factors, for strict paired comparisons. Student fine-tuning updates only the CeNN cell, uses shuffled training examples across the full training set, and stops after a validation plateau. It does not force improvement over a teacher that the compressed baseline already trails.

**Release policy:** Upload a model privately to `vtava/Laya-v54-optimized-liquid-cenn` **only if** full gold validation is not worse than the compressed baseline, the frozen 200-decision TEST is not worse than the same baseline, TEST is no more than 1.01 percentage points below the teacher, fidelity/divergence gates pass, actual checkpoint bytes are reduced from both the original teacher and compressed baseline, speed is at least 0.95× that baseline on tested shapes, and the standalone reload exactly recovers predictions. If speed is below 1× the card will explicitly call it **compression without speedup**.

**Downloads:** The final Colab cell downloads only `Laya_V54_Statistics.zip` (JSON/CSV reports, selection, timing, HF status and checksums). It **never downloads model weights or checkpoint factors to your browser**. If all gates pass, the complete standalone model is uploaded to Hugging Face directly from Colab. Failed models remain local to Colab, never mislabeled as good.

The TEST split has appeared in prior experiments; it is not a never-examined cross-study holdout. No V5.4 empirical performance is claimed before running this notebook.

```
#@title 4. CASE-DISJOINT TRAIN / PROBE / AUDIT benchmark from V4.6
def _jsonish(x):
    if isinstance(x, str):
        try:
            return json.loads(x)
        except Exception:
            return x
    return x

def _canon_label(x):
    s = str(x).strip().lower()
    try:
        v = float(s)
        if math.isfinite(v) and v.is_integer():
            return str(int(v))
    except Exception:
        pass
    return s

def _gold_metadata(q, gold_answer):
    if not isinstance(gold_answer, dict):
        raise ValueError("missing gold answer")
    probs_raw = _jsonish(gold_answer.get("probabilities", {}))
    if not isinstance(probs_raw, dict) or len(probs_raw) < 2:
        raise ValueError("gold probabilities missing or malformed")

    gold_keys = [str(k) for k in probs_raw.keys()]
    if q["t"] == "choice":
        labels = [str(k) for k in q["crit"].keys()]
    elif q["t"] == "noul":
        labels = ["false", "true"]
    else:
        labels = list(gold_keys)
        if all(_canon_label(k).lstrip("-").isdigit() for k in labels):
            labels = sorted(labels, key=lambda k: int(_canon_label(k)))

    lookup = {_canon_label(k): float(v) for k, v in probs_raw.items()}
    if len(labels) != len(gold_keys) or any(_canon_label(k) not in lookup for k in labels):
        raise ValueError("Gold labels do not match encoded option order")

    gold_probs = np.asarray([lookup[_canon_label(k)] for k in labels], dtype=np.float64)
    gold_probs = np.clip(gold_probs, 0.0, None)
    if not np.isfinite(gold_probs).all() or gold_probs.sum() <= 0:
        raise ValueError("invalid gold probability vector")
    gold_probs /= gold_probs.sum()

    gold_label = _canon_label(gold_answer.get("label", labels[int(gold_probs.argmax())]))
    canon_labels = [_canon_label(k) for k in labels]
    if gold_label not in canon_labels:
        raise ValueError(f"gold label {gold_label!r} not in {labels!r}")
    return labels, gold_probs.tolist(), canon_labels.index(gold_label)

def make_items_from_rows(rows, seed, shuffle_items):
    out = []
    head_max_len = int(source_cfg.get("head_max_len", 192))
    for row in rows:
        state = _jsonish(row["state"])
        questions = _jsonish(row["questions"])
        gold = _jsonish(row["gold"])
        case_id = str(row.get("id", ""))
        workflow = str(row.get("workflow", "unknown"))
        state_text = state if isinstance(state, str) else json.dumps(state, ensure_ascii=False)
        state_ids = tokenizer(
            state_text.replace(tokenizer.mask_token, " "),
            add_special_tokens=False,
        )["input_ids"]

        for qid, qdef in questions.items():
            q = normalize_question(qdef)
            ids, markers = build_sequence(
                tokenizer,
                state,
                q,
                max_len=min(MAX_LEN, int(source_cfg.get("max_len", MAX_LEN))),
                head_max_len=head_max_len,
                truncate_left=isinstance(state, list),
                state_ids=state_ids,
            )
            if len(markers) < 2:
                continue
            labels, gold_probs, gold_index = _gold_metadata(q, gold[qid])
            if len(labels) != len(markers):
                raise ValueError(
                    f"{case_id}/{qid}: {len(markers)} markers vs {len(labels)} gold labels"
                )
            out.append({
                "ids": ids,
                "markers": markers,
                "qtype": QTYPES[q["t"]],
                "qtype_name": q["t"],
                "qid": str(qid),
                "case_id": case_id,
                "workflow": workflow,
                "labels": labels,
                "gold_probs": gold_probs,
                "gold_index": int(gold_index),
            })
    if shuffle_items:
        random.Random(seed).shuffle(out)
    return out

# Preserve V5.0's exact case-level TRAIN and PROBE split to reproduce its
# 120/120 step-120 finding. Use the DATASET'S DISTINCT TEST SPLIT for AUDIT.
# Reuse historical V5.1 train/PROBE for comparability, but SELECT using
# unused second-half DEV cases. TEST split stays unseen until after freezing.
rows=list(load_dataset("LocalLLaMA/typed-decisions","all",split="train"))
random.Random(SEED).shuffle(rows)
assert TRAIN_CASES+DEV_CASES<=len(rows)
train_rows=rows[:TRAIN_CASES]
historical_probe_rows=rows[TRAIN_CASES:TRAIN_CASES+DEV_CASES//2]
gold_val_rows=rows[TRAIN_CASES+DEV_CASES//2:TRAIN_CASES+DEV_CASES]
train_items=make_items_from_rows(train_rows,SEED,True)
probe_items=make_items_from_rows(historical_probe_rows,SEED+1,True)
gold_val_items=make_items_from_rows(gold_val_rows,SEED+2,True)
# Case namespaces protect against overlapping local split ids.
test_rows=list(load_dataset("LocalLLaMA/typed-decisions","all",split="test"))
random.Random(SEED+9001).shuffle(test_rows)
audit_rows=[dict(row,id="TEST::"+str(row.get("id",i)))
            for i,row in enumerate(test_rows)]
audit_items=make_items_from_rows(audit_rows,SEED+9001,True)
assert len(probe_items)>=HISTORICAL_PROBE_DECISIONS
assert len(gold_val_items)>=GOLD_VAL_DECISIONS
assert len(audit_items)>=AUDIT_DECISIONS
groups=[{x["case_id"] for x in seq} for seq in
        (train_items,probe_items,gold_val_items,audit_items)]
for i in range(len(groups)):
    for j in range(i+1,len(groups)):
        assert groups[i].isdisjoint(groups[j]),(i,j)
def direct_batch(selected):
    b=collate_items(selected,int(tokenizer.pad_token_id))
    return {k:v.to(device) for k,v in b.items()}
def batch(items,offset,n=BATCH):
    return direct_batch([items[(offset+i)%len(items)] for i in range(n)])
print("TRAIN:",len(train_items),"OLD PROBE:",len(probe_items),
      "NEW GOLD VAL:",len(gold_val_items),"FINAL TEST:",len(audit_items))
```

```
#@title 7. Low-cost 2-stage liquid CeNN layer/width sweep, validated on 180 gold decisions
# Stage 1: 32-channel/2-step replacement for each candidate layer.
# Stage 2: 64-channel/2-step alternatives for top two layers by 48-case screen.
# Screen on TRAIN-captured activations and VALIDATION only; never TEST.
ALL_LAYERS=tuple(range(28)) if SCAN_ALL_28_LAYERS else tuple(CANDIDATE_LAYERS)
assert ALL_LAYERS and len(set(ALL_LAYERS))==len(ALL_LAYERS)
assert all(0<=int(i)<28 for i in ALL_LAYERS)
SCREEN_REF=GOLD_VAL_REF[:SCREEN_VAL_DECISIONS]
SCREEN_ITEMS=gold_val_items[:SCREEN_VAL_DECISIONS]
BASELINE_SCREEN=quality(baseline_model,SCREEN_ITEMS,SCREEN_REF)
scan=[]
stored_candidates={}
def capture_projection_io(layer):
    saved=[]
    target=student.encoder.layers[int(layer)].mlp.Wo
    def capture(_module,args,output):
        saved.append((args[0].detach().to("cpu",dtype=torch.float16).clone(),
                      output.detach().to("cpu",dtype=torch.float16).clone()))
    hook=target.register_forward_hook(capture)
    try:
        for k in range(CAPTURE_BATCHES):
            b=batch(train_items,k*BATCH,BATCH)
            with torch.no_grad(),torch.autocast("cuda",dtype=amp_dtype):
                student(**b)
    finally:
        hook.remove()
    assert len(saved)==CAPTURE_BATCHES
    return target,saved

def fit_candidate(layer,width,measure_full=False):
    original,activations=capture_projection_io(layer)
    try:
        candidate=v53.LiquidCeNNProjection(
            original,width=int(width),steps=LIQUID_STEPS,
            mode="replace",dt_bound=CE_DT_BOUND).to(
                device=DEVICE,dtype=torch.float32)
        old_count=sum(p.numel() for p in original.parameters())
        new_count=sum(p.numel() for p in candidate.parameters())
        if new_count>=old_count:
            record=dict(layer=int(layer),width=int(width),steps=LIQUID_STEPS,
                old_parameters=old_count,cenn_parameters=new_count,
                parameter_saving=old_count-new_count,skip_reason="no_parameter_saving")
            scan.append(record)
            return record
        opt=torch.optim.AdamW(candidate.parameters(),lr=IMITATION_LR,weight_decay=0.)
        losses=[]
        for step in range(1,IMITATION_STEPS+1):
            x0,y0=activations[(step-1)%len(activations)]
            x=x0.to(DEVICE,dtype=torch.float32)
            y=y0.to(DEVICE,dtype=torch.float32)
            candidate.train()
            opt.zero_grad(set_to_none=True)
            pred=candidate(x)
            relative_mse=F.mse_loss(pred,y)/y.square().mean().clamp_min(1e-6)
            assert bool(torch.isfinite(relative_mse))
            relative_mse.backward()
            torch.nn.utils.clip_grad_norm_(candidate.parameters(),1.0)
            opt.step()
            if step==1 or step%20==0 or step==IMITATION_STEPS:
                losses.append(dict(step=step,relative_mse=float(relative_mse.detach().cpu())))
        candidate.eval().requires_grad_(False)
        student.encoder.layers[int(layer)].mlp.Wo=candidate
        try:
            qsmall=quality(student,SCREEN_ITEMS,SCREEN_REF)
            # Full gold validation is mandatory for final selection, but small
            # screens are sufficient to prioritize the 64-channel alternatives.
            qfull=quality(student,gold_val_items,GOLD_VAL_REF) if measure_full else None
        finally:
            student.encoder.layers[int(layer)].mlp.Wo=original
        entry=dict(layer=int(layer),width=int(width),steps=LIQUID_STEPS,
            old_parameters=old_count,cenn_parameters=new_count,
            parameter_saving=old_count-new_count,
            screen_quality=qsmall,full_gold_validation=qfull,
            last_relative_mse=losses[-1]["relative_mse"],imitation_losses=losses)
        scan.append(entry)
        stored_candidates[(int(layer),int(width))]={
            k:t.detach().cpu().contiguous().clone() for k,t in candidate.state_dict().items()}
        print(f"LIQUID layer={layer:02d} width={width:03d} steps={LIQUID_STEPS} "
              f"saved={(old_count-new_count)*4/1e6:.2f} MB "
              f"screenGold={qsmall['gold_accuracy']:.3f} "
              f"JS={qsmall['mean_js']:.5f}")
        return entry
    finally:
        del activations

for layer in ALL_LAYERS:
    fit_candidate(layer,FAST_WIDTH,measure_full=False)
    (OUT/"liquid_candidate_search.json").write_text(json.dumps(scan,indent=2))
fast_ok=[r for r in scan if "screen_quality" in r]
assert fast_ok,"No parameter-saving liquid replacement candidate"
leaders=sorted(fast_ok,key=lambda r:(
    -r["screen_quality"]["gold_accuracy"],
    r["screen_quality"]["mean_js"],
    -r["parameter_saving"]))[:2]
if WIDE_WIDTH!=FAST_WIDTH:
    for r in leaders:
        fit_candidate(r["layer"],WIDE_WIDTH,measure_full=False)
        (OUT/"liquid_candidate_search.json").write_text(json.dumps(scan,indent=2))

# Full 180-decision validation for every candidate that preserved gold
# correctness on the small screen or belonged to the 2 finalists.
shortlisted=[r for r in scan if "screen_quality" in r and (
    r["screen_quality"]["gold_accuracy"]>=
        BASELINE_SCREEN["gold_accuracy"]-1./SCREEN_VAL_DECISIONS
    or r["layer"] in [x["layer"] for x in leaders])]
assert shortlisted
for r in shortlisted:
    layer,width=r["layer"],r["width"]
    original=v53.install_liquid_ffn(student,layer,width=width,
        steps=LIQUID_STEPS,mode="replace",dt_bound=CE_DT_BOUND)
    cell=student.encoder.layers[layer].mlp.Wo
    cell.load_state_dict(stored_candidates[(layer,width)],strict=True)
    try:
        r["full_gold_validation"]=quality(student,gold_val_items,GOLD_VAL_REF)
    finally:
        v53.uninstall_liquid_ffn(student,layer,original)
    (OUT/"liquid_candidate_search.json").write_text(json.dumps(scan,indent=2))
    print("FULL VALIDATION",layer,width,r["full_gold_validation"])

# Gold correctness FIRST, teacher decision agreement SECOND, JS THIRD,
# memory saving FOURTH; no selection on untouched TEST labels.
evaluated=[r for r in shortlisted if r.get("full_gold_validation") is not None]
assert evaluated
def candidate_score(r):
    q=r["full_gold_validation"]
    return (q["gold_accuracy"],q["teacher_agreement"],
            -q["mean_js"],r["parameter_saving"])
best_candidate=max(evaluated,key=candidate_score)
SELECTED_LAYER=int(best_candidate["layer"])
SELECTED_WIDTH=int(best_candidate["width"])
SELECTED_MODE="replace"
original_ffn=v53.install_liquid_ffn(
    student,SELECTED_LAYER,width=SELECTED_WIDTH,
    steps=LIQUID_STEPS,mode=SELECTED_MODE,dt_bound=CE_DT_BOUND)
liquid_cell=student.encoder.layers[SELECTED_LAYER].mlp.Wo
liquid_cell.load_state_dict(
    stored_candidates[(SELECTED_LAYER,SELECTED_WIDTH)],strict=True)
del stored_candidates
gc.collect()
LIQUID_CONFIG=dict(layer=SELECTED_LAYER,mode=SELECTED_MODE,
    width=SELECTED_WIDTH,steps=LIQUID_STEPS,dt_bound=CE_DT_BOUND)
initial_liquid=quality(student,gold_val_items,GOLD_VAL_REF)
assert initial_liquid==best_candidate["full_gold_validation"]
SELECTED_FFN_SAVING=int(best_candidate["parameter_saving"])*4
assert SELECTED_FFN_SAVING>0
(OUT/"selected_liquid_architecture.json").write_text(json.dumps({
    "liquid_cell":LIQUID_CONFIG,"best_candidate":best_candidate,
    "baseline_gold_validation":BASELINE_VAL,
    "selected_validation":initial_liquid,
    "saved_fp32_bytes_of_replaced_projection":SELECTED_FFN_SAVING,
    "selection_metric":"full gold accuracy > agreement > JS > storage"
},indent=2))
print("SELECTED OPTIMIZED LIQUID",LIQUID_CONFIG,
      "saved FNN projection bytes",SELECTED_FFN_SAVING)
```

```
#@title 8. Shuffled gold-aware CeNN fine-tune, preserve highest gold-validation accuracy
# All 28 encoder layers remain present; no original pretrained weight updates.
student.requires_grad_(False)
liquid_cell=student.encoder.layers[SELECTED_LAYER].mlp.Wo
for name,p in liquid_cell.named_parameters():
    # Residual base FFN is FROZEN; only liquid dynamics are trainable.
    if not name.startswith("base."):
        p.requires_grad_(True)
if TRAIN_DECISION_SCORER:
    for p in student.scorer.parameters(): p.requires_grad_(True)
trainable=[p for p in student.parameters() if p.requires_grad]
assert trainable
assert not any(p.requires_grad for p in teacher.parameters())
trainable_specs={name:list(p.shape)
      for name,p in student.named_parameters() if p.requires_grad}
trainable_names=set(trainable_specs)
trainable_fingerprint=hashlib.sha256(json.dumps(
       trainable_specs,sort_keys=True).encode()).hexdigest()
student.eval()
BEST=OUT/"best_liquid_trainable.safetensors"
HISTORY=OUT/"liquid_training_progress.json"
def save_selected(step,quality_value,reason):
    from collections import OrderedDict
    state=OrderedDict((name,p.detach().cpu().contiguous().clone())
      for name,p in student.named_parameters() if p.requires_grad)
    save_file(state,str(BEST),metadata={
      "format":"pt","fingerprint":trainable_fingerprint,"step":str(step)})
    (OUT/"best_liquid_trainable.json").write_text(json.dumps(
       dict(step=step,quality=quality_value,reason=reason,
            trainable_fingerprint=trainable_fingerprint,
            liquid_config=LIQUID_CONFIG),indent=2))
def restore_selected():
    weights=load_file(str(BEST),device="cpu")
    assert set(weights)==trainable_names
    with torch.no_grad():
        for name,p in student.named_parameters():
            if name in weights:
                p.copy_(weights[name].to(p.device,p.dtype))
def rank_gold(q):
    return (q["gold_accuracy"],
            -max(0.,q["teacher_gold_accuracy"]-q["gold_accuracy"]),
            -q["mean_js"])
# Deterministic epoch-shuffled coverage instead of always training on the
# first N examples when early stopping occurs. Two examples/step can traverse
# all 500 training cases within 250 steps.
train_order=list(range(len(train_items)))
train_rng=random.Random(SEED+541)
train_rng.shuffle(train_order)
train_cursor=0
def make_gold_batch(step):
    global train_cursor
    choices=[]
    for _ in range(TRAIN_B):
        if train_cursor>=len(train_order):
            train_rng.shuffle(train_order)
            train_cursor=0
        choices.append(train_order[train_cursor])
        train_cursor+=1
    entries=[train_items[k] for k in choices]
    b=direct_batch(entries)
    assert bool((b["marker_pos"][b["marker_mask"]]<TRAIN_MAX_TOKENS).all())
    for key in ("input_ids","attention_mask"):
        b[key]=b[key][:,:TRAIN_MAX_TOKENS]
    gold=torch.tensor([item["gold_index"] for item in entries],
        dtype=torch.long,device=DEVICE)
    soft=torch.zeros(b["marker_mask"].shape,dtype=torch.float32,device=DEVICE)
    for i,item in enumerate(entries):
        goldp=torch.tensor(item["gold_probs"],dtype=torch.float32,device=DEVICE)
        soft[i,:len(goldp)]=goldp/goldp.sum().clamp_min(1e-8)
    return b,gold,soft
best_score=quality(student,gold_val_items,GOLD_VAL_REF)
best_step=0
save_selected(0,best_score,"start_from_imitation_or_identity")
training_history=[]
optimizer=torch.optim.AdamW(trainable,lr=TRAIN_LR,weight_decay=0.)
scaler=torch.amp.GradScaler("cuda",enabled=amp_dtype==torch.float16)
patience_checks=0
for step in range(1,TRAIN_STEPS+1):
    b,gold,soft=make_gold_batch(step)
    with torch.no_grad(),torch.autocast("cuda",dtype=amp_dtype):
        teacher_logits,_=teacher(**b)
    optimizer.zero_grad(set_to_none=True)
    with torch.autocast("cuda",dtype=amp_dtype):
        scores,_=student(**b)
    valid=b["marker_mask"].bool()
    s=scores.float().masked_fill(~valid,-1e9)
    t=teacher_logits.float().masked_fill(~valid,-1e9)
    logs=F.log_softmax(s,dim=-1)
    ce=F.nll_loss(logs,gold)
    soft_ce=-(soft*logs).sum(-1).mean()
    kl=F.kl_div(F.log_softmax(s/2.,dim=-1),
       F.softmax(t/2.,dim=-1),reduction="batchmean")*4.
    loss=GOLD_CE_WEIGHT*ce+GOLD_SOFT_WEIGHT*soft_ce+TEACHER_KL_WEIGHT*kl
    if not bool(torch.isfinite(loss)):
        raise RuntimeError(f"Non-finite CeNN training loss at {step}")
    scaler.scale(loss).backward()
    scaler.unscale_(optimizer)
    torch.nn.utils.clip_grad_norm_(trainable,1.)
    scaler.step(optimizer);scaler.update()
    if step%TRAIN_CHECK_EVERY==0 or step==TRAIN_STEPS:
        q=quality(student,gold_val_items,GOLD_VAL_REF)
        improved=rank_gold(q)>rank_gold(best_score)
        if improved:
            best_score=q;best_step=step
            save_selected(step,q,"best_gold_validation")
            patience_checks=0
        else:
            patience_checks+=1
        training_history.append(dict(step=step,
              loss=float(loss.detach().cpu()),quality=q,
              gold_gain_vs_teacher_pp=100*(
                  q["gold_accuracy"]-q["teacher_gold_accuracy"]),
              improved=improved))
        HISTORY.write_text(json.dumps(training_history,indent=2))
        print(f"step={step} mode={SELECTED_MODE} "
             f"gold={q['gold_accuracy']:.3%} "
             f"teacher={q['teacher_gold_accuracy']:.3%} "
             f"JS={q['mean_js']:.6f} selected={improved}")
        if patience_checks>=4:
            print("EARLY STOP: 4 validation checks without improved gold correctness")
            break
restore_selected()
student.eval().requires_grad_(False)
FINAL_VAL=quality(student,gold_val_items,GOLD_VAL_REF)
HISTORICAL_PROBE=quality(student,probe_items,PROBE_REF)
assert FINAL_VAL["gold_accuracy"]==best_score["gold_accuracy"]
(OUT/"liquid_training_summary.json").write_text(json.dumps(dict(
    liquid_config=LIQUID_CONFIG,control_validation=BASELINE_VAL,
    selected_step=best_step,final_validation=FINAL_VAL,
    historical_probe=HISTORICAL_PROBE,
    training_history=training_history,
    trainable_parameter_count=sum(p.numel() for p in trainable)
),indent=2))
print("CeNN validation:",FINAL_VAL,"selected step:",best_step)
```

```
#@title 9. FINAL frozen TEST evaluation vs EXACT compressed baseline and teacher
assert not any(p.requires_grad for p in student.parameters())
# baseline_model is an exact model created once from state_dict in cell 6.
assert v53.unique_parameter_count(baseline_model)==BASELINE_UNIQUE
BENCH_ORIGINAL=BENCH_BATCHES
BENCH_BATCHES=[batch(audit_items,0,BATCH),batch(audit_items,11,BATCH)]
TEST_REF=get_reference(audit_items,AUDIT_DECISIONS)
assert {x["case"] for x in TEST_REF}.isdisjoint(
    {x["case"] for x in GOLD_VAL_REF})
test_quality=quality(student,audit_items,TEST_REF)
baseline_test=quality(baseline_model,audit_items,TEST_REF)
speed_against_baseline=benchmark_paired(
    baseline_model,student,repeats=max(BENCH_REPEATS,15))
speed_against_teacher=benchmark_paired(
    teacher,student,repeats=max(BENCH_REPEATS,15))
BENCH_BATCHES=BENCH_ORIGINAL

@torch.inference_mode()
def disagreement_report(items,ref):
    from collections import Counter
    count=Counter()
    by_kind={}
    for i in range(0,len(ref),BATCH):
        records=items[i:i+BATCH]
        p=probs_for(student,direct_batch(records))
        for j,item in enumerate(records):
            actual=int(p[j].argmax())
            truth=int(item["gold_index"])
            source=int(ref[i+j]["teacher"])
            if source!=truth and actual==truth: label="corrected_teacher_error"
            elif source==truth and actual!=truth: label="introduced_error"
            elif source==truth: label="both_correct"
            else: label="both_wrong"
            count[label]+=1
            by_kind.setdefault(item["qtype_name"],Counter())[label]+=1
    return dict(overall=dict(count),
        by_type={k:dict(v) for k,v in by_kind.items()})
test_errors=disagreement_report(audit_items,TEST_REF)
audit_report=dict(
    teacher_gold=test_quality["teacher_gold_accuracy"],
    baseline_gold=baseline_test["gold_accuracy"],
    liquid_gold=test_quality["gold_accuracy"],
    gold_difference_pp=100*(test_quality["gold_accuracy"]-
        test_quality["teacher_gold_accuracy"]),
    gold_gain_vs_baseline_pp=100*(test_quality["gold_accuracy"]-
        baseline_test["gold_accuracy"]),
    liquid_test_metrics=test_quality,
    compressed_baseline_test_metrics=baseline_test,
    error_breakdown=test_errors,
    gpu_wall_speed_vs_baseline=speed_against_baseline,
    gpu_wall_speed_vs_teacher=speed_against_teacher
)
(OUT/"liquid_test_report.json").write_text(json.dumps(audit_report,indent=2))
print("FROZEN TEST",json.dumps({
    "teacher_gold":audit_report["teacher_gold"],
    "baseline_gold":audit_report["baseline_gold"],
    "liquid_gold":audit_report["liquid_gold"],
    "liquid_vs_baseline_pp":audit_report["gold_gain_vs_baseline_pp"],
    "liquid_vs_teacher_pp":audit_report["gold_difference_pp"],
    "gpu_ratio_vs_baseline":speed_against_baseline["median_gpu_speedup"]
},indent=2))
```
