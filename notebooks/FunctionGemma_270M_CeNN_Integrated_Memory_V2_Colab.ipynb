# FunctionGemma 270M × TinyCeNN — Integrated Memory V2

Corrected experiment for `vtava/functiongemma-270m-it-simple-tool-calling`.

Gemma3 has 18 decoder layers with full/global attention at **5, 11, 17** and native sliding attention elsewhere.

V2 fixes the cache-equivalence diagnostic: cached/full top-1 drift is judged **relative to untouched FunctionGemma on the same tokens**, not against an inappropriate fixed 98% threshold.


```python
import os, sys, json, subprocess, tempfile, shutil
from pathlib import Path
from datetime import datetime, timezone

REPO = Path(tempfile.mkdtemp(prefix="functiongemma-cenn-v2-")) / "TinyCeNN-LM"
subprocess.run(["git","clone","--quiet","https://github.com/vtavakkoli/TinyCeNN-LM.git",str(REPO)], check=True)
subprocess.run(["git","checkout","main"], cwd=REPO, check=True)
subprocess.run(["git","fetch","origin","main"], cwd=REPO, check=True)
subprocess.run(["git","reset","--hard","origin/main"], cwd=REPO, check=True)

subprocess.run([sys.executable,"-m","pip","install","-q",
                "transformers==4.57.6","huggingface_hub==0.36.2","datasets>=3,<5",
                "pytest","pandas","matplotlib"], check=True)
subprocess.run([sys.executable,"-m","pip","install","-q","-e",str(REPO),"--no-deps"], check=True)

os.environ["PYTHONPATH"] = os.pathsep.join([str(REPO),str(REPO/"src")])
sys.path[:0] = [str(REPO),str(REPO/"src")]

import torch
from transformers import AutoConfig
MODEL_ID = "vtava/functiongemma-270m-it-simple-tool-calling"
cfg = AutoConfig.from_pretrained(MODEL_ID).get_text_config(decoder=True)
FULL = [i for i,t in enumerate(cfg.layer_types) if t == "full_attention"]
print("Source:",subprocess.check_output(["git","rev-parse","HEAD"],cwd=REPO,text=True).strip())
print("GPU:",torch.cuda.get_device_name(0) if torch.cuda.is_available() else "CPU")
print("Architecture:",cfg.model_type,"layers:",cfg.num_hidden_layers,"full attention:",FULL,"window:",cfg.sliding_window)
assert cfg.model_type == "gemma3_text" and FULL == [5,11,17]
```

## Train + evaluate with the corrected V2 cache gate


```python
BENCHMARK = REPO / "scripts/benchmark_functiongemma_integrated_memory_v2.py"
cmd = [sys.executable,"-u",str(BENCHMARK),"--base-model",MODEL_ID,"--output-dir",str(OUT)]
for k,v in RUN.items():
    cmd += ["--" + k.replace("_","-"), str(v)]
print("Running V2 benchmark:",BENCHMARK.name)
print(" ".join(cmd))
try:
    with LOG.open("w") as log:
        with subprocess.Popen(cmd,cwd=REPO,stdout=subprocess.PIPE,stderr=subprocess.STDOUT,text=True,bufsize=1) as p:
            for line in p.stdout:
                print(line,end="",flush=True)
                log.write(line); log.flush()
            status = p.wait()
    if status:
        raise RuntimeError(f"Run failed: {status}; inspect {LOG}")
finally:
    if OUT.exists() and LOG.exists():
        shutil.copy2(LOG,OUT/"console.log")
        print("Archive:",shutil.make_archive(str(OUT)+"-results","zip",root_dir=OUT))
```

## Tool-calling preservation — original vs selected CeNN


```python
import re
from transformers import AutoModelForCausalLM,AutoTokenizer,DynamicCache
from tinycenn_lm.gemma3_integrated_memory import restore_student,inference_mode,greedy_generate,native_dtype
M=json.loads((OUT/"manifest.json").read_text()); R=json.loads((OUT/"integrated_report.json").read_text()); S=json.loads((OUT/"selection.json").read_text())["selected"]; rec=next(r for r in R["candidates"] if r["candidate"]==S)
dev=torch.device("cuda" if torch.cuda.is_available() else "cpu"); dt=native_dtype(dev)
tok=AutoTokenizer.from_pretrained(MODEL_ID,revision=M["model_revision"])
teacher=AutoModelForCausalLM.from_pretrained(MODEL_ID,revision=M["model_revision"],dtype=dt,attn_implementation="sdpa").to(dev).eval()
student=restore_student(teacher,torch.load(OUT/rec["checkpoint"],map_location="cpu",weights_only=True)).to(dev).eval()
@torch.no_grad()
def gen_original(ids,n=48):
    cache=DynamicCache(config=teacher.config); pieces=[]
    token=teacher(input_ids=ids,past_key_values=cache,use_cache=True).logits[:,-1].argmax(-1,keepdim=True)
    for _ in range(n):
        pieces.append(token); token=teacher(input_ids=token,past_key_values=cache,use_cache=True).logits[:,-1].argmax(-1,keepdim=True)
    return torch.cat(pieces,1)
tools=[{"type":"function","function":{"name":"get_current_temperature","description":"Get temperature for a city","parameters":{"type":"object","properties":{"location":{"type":"string"}},"required":["location"]}}},{"type":"function","function":{"name":"calculate","description":"Calculate an expression","parameters":{"type":"object","properties":{"expression":{"type":"string"}},"required":["expression"]}}}]
for prompt in ["What's the temperature in Vienna?","Calculate 18 times 7."]:
    messages=[{"role":"developer","content":"You are a model that can do function calling with the following functions"},{"role":"user","content":prompt}]
    ids=tok.apply_chat_template(messages,tools=tools,add_generation_prompt=True,tokenize=True,return_dict=True,return_tensors="pt")["input_ids"].to(dev)
    Aids=gen_original(ids)
    with inference_mode(student,str(dt).removeprefix("torch.")): Bids,_=greedy_generate(student,ids,48)
    A=tok.decode(Aids[0],skip_special_tokens=False); B=tok.decode(Bids[0],skip_special_tokens=False)
    pattern=r"call:([A-Za-z_][A-Za-z0-9_]*)"
    def fname(text):
        m=re.search(pattern,text); return m.group(1) if m else None
    print("\n"+"="*90); print("PROMPT:",prompt); print("ORIGINAL:",A); print("CeNN:",B); print("FUNCTION:",fname(A),"vs",fname(B))
```
