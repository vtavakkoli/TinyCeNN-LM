### What V5.1 changes

1. Reuse exactly the **20 selected projections and ranks** from your supplied `greedy_combinations.json`. Recreate factor initialization with V5.0's deterministic per-family SVD seed. No expensive brute-force rerun.
2. **Check PROBE every 10 steps** (configurable). Prefer teacher agreement first, then gold preservation, then JS. Save the best trainable A/B tensors *to disk*.
3. **STOP IMMEDIATELY** on 100% top-1 teacher agreement **and** no gold drop **and** JS ≤0.003 (confirmed by a second evaluation at the same weights). Max 240 steps; stop before 240 when conditions are met.
4. Use the **separate test split** of `LocalLLaMA/typed-decisions` as a new 200-decision AUDIT, not the previously reused training/dev split. Evaluate it once after freezing model. Test results are independently gated.
5. Export a **complete FP32 model** with offline loader, tokenizer, actual byte report, reproducibility manifest, and strict offline reload parity.
6. Hugging Face publication is **conditional** on fidelity, storage and reload gates. Paired CUDA/wall speed is always measured and disclosed; optional strict speed-gating supports compression-only publication without claiming a speedup.

**Important:** A perfect *decision match* means identical argmax options on the 120-decision PROBE, not identical logits/probabilities or universal 100% accuracy. Results from V5.0 are observations; V5.1 has not been trained until this Colab executes.

**Separate Hugging Face target:** `vtava/Laya-v51-all28-earlystop-standalone` (private by default). No existing model will be overwritten automatically.

````python
#@title 4. CASE-DISJOINT TRAIN / PROBE / AUDIT benchmark from V4.6
def _jsonish(x):
    if isinstance(x, str):
        try:
            return json.loads(x)
        except Exception:
            return x
    return x

def _canon_label(x):
    s = str(x).strip().lower()
    try:
        v = float(s)
        if math.isfinite(v) and v.is_integer():
            return str(int(v))
    except Exception:
        pass
    return s

def _gold_metadata(q, gold_answer):
    if not isinstance(gold_answer, dict):
        raise ValueError("missing gold answer")
    probs_raw = _jsonish(gold_answer.get("probabilities", {}))
    if not isinstance(probs_raw, dict) or len(probs_raw) < 2:
        raise ValueError("gold probabilities missing or malformed")

    gold_keys = [str(k) for k in probs_raw.keys()]
    if q["t"] == "choice":
        labels = [str(k) for k in q["crit"].keys()]
    elif q["t"] == "noul":
        labels = ["false", "true"]
    else:
        labels = list(gold_keys)
        if all(_canon_label(k).lstrip("-").isdigit() for k in labels):
            labels = sorted(labels, key=lambda k: int(_canon_label(k)))

    lookup = {_canon_label(k): float(v) for k, v in probs_raw.items()}
    if len(labels) != len(gold_keys) or any(_canon_label(k) not in lookup for k in labels):
        raise ValueError("Gold labels do not match encoded option order")

    gold_probs = np.asarray([lookup[_canon_label(k)] for k in labels], dtype=np.float64)
    gold_probs = np.clip(gold_probs, 0.0, None)
    if not np.isfinite(gold_probs).all() or gold_probs.sum() <= 0:
        raise ValueError("invalid gold probability vector")
    gold_probs /= gold_probs.sum()

    gold_label = _canon_label(gold_answer.get("label", labels[int(gold_probs.argmax())]))
    canon_labels = [_canon_label(k) for k in labels]
    if gold_label not in canon_labels:
        raise ValueError(f"gold label {gold_label!r} not in {labels!r}")
    return labels, gold_probs.tolist(), canon_labels.index(gold_label)

def make_items_from_rows(rows, seed, shuffle_items):
    out = []
    head_max_len = int(source_cfg.get("head_max_len", 192))
    for row in rows:
        state = _jsonish(row["state"])
        questions = _jsonish(row["questions"])
        gold = _jsonish(row["gold"])
        case_id = str(row.get("id", ""))
        workflow = str(row.get("workflow", "unknown"))
        state_text = state if isinstance(state, str) else json.dumps(state, ensure_ascii=False)
        state_ids = tokenizer(
            state_text.replace(tokenizer.mask_token, " "),
            add_special_tokens=False,
        )["input_ids"]

        for qid, qdef in questions.items():
            q = normalize_question(qdef)
            ids, markers = build_sequence(
                tokenizer,
                state,
                q,
                max_len=min(MAX_LEN, int(source_cfg.get("max_len", MAX_LEN))),
                head_max_len=head_max_len,
                truncate_left=isinstance(state, list),
                state_ids=state_ids,
            )
            if len(markers) < 2:
                continue
            labels, gold_probs, gold_index = _gold_metadata(q, gold[qid])
            if len(labels) != len(markers):
                raise ValueError(
                    f"{case_id}/{qid}: {len(markers)} markers vs {len(labels)} gold labels"
                )
            out.append({
                "ids": ids,
                "markers": markers,
                "qtype": QTYPES[q["t"]],
                "qtype_name": q["t"],
                "qid": str(qid),
                "case_id": case_id,
                "workflow": workflow,
                "labels": labels,
                "gold_probs": gold_probs,
                "gold_index": int(gold_index),
            })
    if shuffle_items:
        random.Random(seed).shuffle(out)
    return out

# Preserve V5.0's exact case-level TRAIN and PROBE split to reproduce its
# 120/120 step-120 finding. Use the DATASET'S DISTINCT TEST SPLIT for AUDIT.
rows=list(load_dataset("LocalLLaMA/typed-decisions","all",split="train"))
rng=random.Random(SEED)
rng.shuffle(rows)
assert TRAIN_CASES+DEV_CASES<=len(rows)
train_rows=rows[:TRAIN_CASES]
probe_rows=rows[TRAIN_CASES:TRAIN_CASES+DEV_CASES//2]
train_items=make_items_from_rows(train_rows,SEED,True)
probe_items=make_items_from_rows(probe_rows,SEED+1,True)
test_rows=list(load_dataset("LocalLLaMA/typed-decisions","all",split="test"))
random.Random(SEED+9001).shuffle(test_rows)
# Namespace case IDs to prevent accidental equality between split-local IDs.
audit_rows=[dict(r,id="TEST::"+str(r.get("id",i)))
            for i,r in enumerate(test_rows)]
audit_items=make_items_from_rows(audit_rows,SEED+9001,True)
assert train_items and probe_items and audit_items
assert len(probe_items)>=PROBE_DECISIONS and len(audit_items)>=AUDIT_DECISIONS
assert {x["case_id"] for x in probe_items}.isdisjoint(
       {x["case_id"] for x in audit_items})
assert {x["case_id"] for x in train_items}.isdisjoint(
       {x["case_id"] for x in probe_items})
def direct_batch(selected):
    b=collate_items(selected,int(tokenizer.pad_token_id))
    return {k:v.to(device) for k,v in b.items()}
def batch(items,offset,n=BATCH):
    return direct_batch([items[(offset+i)%len(items)] for i in range(n)])
print("TRAIN:",len(train_items),"PROBE:",len(probe_items),
      "NEW DATASET TEST AUDIT:",len(audit_items))
````

````python
#@title 5. Accuracy, divergence and paired CUDA/wall timing
@torch.inference_mode()
def probs_for(model,b):
    with torch.autocast("cuda",dtype=amp_dtype):
        scores,_=model(**b)
    return torch.softmax(scores.float().masked_fill(
        ~b["marker_mask"].bool(),-1e9),dim=-1).cpu()

@torch.inference_mode()
def get_reference(items,limit):
    out=[]
    for start in range(0,min(limit,len(items)),BATCH):
        chosen=items[start:min(start+BATCH,limit)]
        p=probs_for(teacher,direct_batch(chosen))
        for j,item in enumerate(chosen):
            out.append(dict(probs=p[j],teacher=int(p[j].argmax()),
                gold=int(item["gold_index"]),case=item["case_id"],qid=item["qid"]))
    return out

@torch.inference_mode()
def quality(model,items,ref):
    agree=correct=tcorrect=0;js=0.
    n=len(ref)
    for start in range(0,n,BATCH):
        chosen=items[start:min(n,start+BATCH)]
        p=probs_for(model,direct_batch(chosen))
        for j,item in enumerate(chosen):
            t=ref[start+j]["probs"]; s=p[j]
            assert s.shape==t.shape
            m=(s+t)/2
            js+=float(.5*((s*(s.clamp_min(1e-8).log()-m.clamp_min(1e-8).log())).sum()+
                            (t*(t.clamp_min(1e-8).log()-m.clamp_min(1e-8).log())).sum()))
            guess=int(s.argmax())
            agree+=int(guess==ref[start+j]["teacher"])
            correct+=int(guess==item["gold_index"])
            tcorrect+=int(ref[start+j]["teacher"]==item["gold_index"])
    return dict(decisions=n,teacher_agreement=agree/n,gold_accuracy=correct/n,
        teacher_gold_accuracy=tcorrect/n,gold_drop=(tcorrect-correct)/n,mean_js=js/n)

BENCH_BATCHES=[batch(probe_items,0,BATCH),batch(probe_items,11,BATCH)]
@torch.inference_mode()
def benchmark_paired(reference,candidate,repeats=BENCH_REPEATS):
    """Alternating A/B/B/A with CUDA events AND synchronized wall-clock timings.
    All timings exclude Colab compilation, dataset loading and model downloading."""
    results=[]
    for j,b in enumerate(BENCH_BATCHES):
        def once(m):
            torch.cuda.synchronize()
            a=torch.cuda.Event(enable_timing=True)
            z=torch.cuda.Event(enable_timing=True)
            before=time.perf_counter()
            a.record()
            with torch.autocast("cuda",dtype=amp_dtype):
                outputs=m(**b)
            z.record();z.synchronize()
            wall=(time.perf_counter()-before)*1000
            return float(a.elapsed_time(z)),float(wall)
        for k in range(4):
            once(reference);once(candidate)
        d={"reference":[],"candidate":[]}
        w={"reference":[],"candidate":[]}
        for k in range(repeats):
            order=("reference","candidate","candidate","reference") if k%2==0 else (
                "candidate","reference","reference","candidate")
            for name in order:
                gpu,cpu=once(reference if name=="reference" else candidate)
                d[name].append(gpu);w[name].append(cpu)
        ref=float(np.median(d["reference"]))
        cand=float(np.median(d["candidate"]))
        rw=float(np.median(w["reference"]))
        cw=float(np.median(w["candidate"]))
        results.append(dict(shape=j,batch=int(b["input_ids"].shape[0]),
            tokens=int(b["input_ids"].shape[1]),
            native_gpu_ms=ref,candidate_gpu_ms=cand,gpu_speedup=ref/max(cand,1e-9),
            native_wall_ms=rw,candidate_wall_ms=cw,wall_speedup=rw/max(cw,1e-9),
            native_p95_ms=float(np.percentile(d["reference"],95)),
            candidate_p95_ms=float(np.percentile(d["candidate"],95)),n=len(d["reference"])))
    return dict(median_gpu_speedup=float(np.median([r["gpu_speedup"] for r in results])),
        min_gpu_speedup=float(min(r["gpu_speedup"] for r in results)),
        median_wall_speedup=float(np.median([r["wall_speedup"] for r in results])),
        min_wall_speedup=float(min(r["wall_speedup"] for r in results)),
        per_shape=results)

def valid_quality(q):
    return (q["teacher_agreement"]>=MIN_TEACHER_AGREEMENT and
            q["gold_drop"]<=MAX_GOLD_DROP_PP/100 and q["mean_js"]<=MAX_JS)
def valid_speed(s):
    return (s["min_gpu_speedup"]>=MIN_GPU_SPEED_RATIO and
            s["min_wall_speedup"]>=MIN_WALL_SPEED_RATIO)

````

````python
#@title 7. Fine-tune selected A/B factors — STOP on first verified 100% teacher match
student.requires_grad_(False)
modules=v50.applied_adapters(student,SELECTED_SPECS)
trainable=[]
for module in modules:
    trainable.extend((module.A,module.B))
for p in trainable:
    p.requires_grad_(True)
assert len(trainable)==40
assert sum(p.numel() for p in student.parameters() if p.requires_grad)==sum(
    p.numel() for p in trainable)
student.eval()
BEST_FILE=OUT/"best_selected_adapters.safetensors"
BEST_METADATA=OUT/"best_selected_adapters.json"
LOG=OUT/"training_progress.json"

def exact_perfect(q):
    # Top-1 agreement must match ALL 120 teacher decisions; no gold drop and
    # no excessive probability drift are permitted for early stopping.
    return (q["teacher_agreement"]>=1.0-1e-12 and
            q["gold_drop"]<=1e-12 and q["mean_js"]<=MAX_JS)
def quality_pass(q):
    return (q["teacher_agreement"]>=MIN_TEACHER_AGREEMENT and
            q["gold_drop"]<=MAX_GOLD_DROP_PP/100 and q["mean_js"]<=MAX_JS)
def rank_checkpoint(q):
    # Perfect decision agreement is ALWAYS preferable to a slightly lower JS
    # that sacrifices one or more predictions (V5.0's step-240 mistake).
    return (int(quality_pass(q)),int(exact_perfect(q)),
            float(q["teacher_agreement"]),
            -max(0.,float(q["gold_drop"])),
            -float(q["mean_js"]))
def adapters_on_cpu():
    tensors=OrderedDict()
    for i,m in enumerate(v50.applied_adapters(student,SELECTED_SPECS)):
        tensors[f"adapter_{i:02d}.A"]=m.A.detach().cpu().contiguous().clone()
        tensors[f"adapter_{i:02d}.B"]=m.B.detach().cpu().contiguous().clone()
    return tensors
def restore_from_weights(path):
    data=load_file(str(path),device="cpu")
    modules=v50.applied_adapters(student,SELECTED_SPECS)
    required={f"adapter_{i:02d}.{k}" for i in range(len(modules))
              for k in ("A","B")}
    assert set(data)==required,(required-set(data),set(data)-required)
    with torch.no_grad():
        for i,m in enumerate(modules):
            for key in ("A","B"):
                name=f"adapter_{i:02d}.{key}"
                dest=getattr(m,key)
                assert dest.shape==data[name].shape,name
                dest.copy_(data[name].to(device=dest.device,dtype=dest.dtype))
def save_best(step,q,reason):
    save_file(adapters_on_cpu(),str(BEST_FILE),metadata={
        "format":"pt","source_revision":PINNED_TEACHER_REVISION,
        "selected_spec_sha256":spec_hash,
        "step":str(step),"reason":reason})
    BEST_METADATA.write_text(json.dumps(dict(
        step=step,quality=q,reason=reason,source_revision=PINNED_TEACHER_REVISION,
        selected_spec_sha256=spec_hash),indent=2),encoding="utf-8")

spec_hash=hashlib.sha256(json.dumps(
    SELECTED_SPECS,sort_keys=True,separators=(",",":")).encode()).hexdigest()
if RESUME_BEST_ADAPTERS_FILE.strip():
    provided=Path(RESUME_BEST_ADAPTERS_FILE.strip()).expanduser()
    assert provided.is_file(),f"Checkpoint missing: {provided}"
    # A matching adjacent .json manifest is mandatory for resume safety.
    manifest=provided.with_suffix(".json")
    assert manifest.is_file(),"Resume requires matching .json metadata"
    meta=json.loads(manifest.read_text())
    assert meta["source_revision"]==PINNED_TEACHER_REVISION
    assert meta["selected_spec_sha256"]==spec_hash
    restore_from_weights(provided)
    print("Restored prior selected A/B state:",provided)

starting=quality(student,probe_items,PROBE_REF)
best_quality=starting
best_step=0
save_best(0,starting,"reconstructed_or_resumed_start")
early_stop_reason=None
training_log=[]
# If the reconstructed/resumed model ALREADY meets the perfect gate, avoid
# unnecessary optimizer steps.
if STOP_ON_PERFECT_MATCH and exact_perfect(starting):
    recheck=quality(student,probe_items,PROBE_REF)
    if exact_perfect(recheck):
        early_stop_reason="perfect_probe_at_initialization"

if early_stop_reason is None and MAX_TRAIN_STEPS>0:
    optimizer=torch.optim.AdamW(trainable,lr=TRAIN_LR,weight_decay=0.)
    scaler=torch.amp.GradScaler("cuda",enabled=amp_dtype==torch.float16)
    for step in range(1,MAX_TRAIN_STEPS+1):
        b=batch(train_items,((step-1)*TRAIN_B)%len(train_items),TRAIN_B)
        assert bool((b["marker_pos"][b["marker_mask"]]<TRAIN_MAX_TOKENS).all()),(
            "Increase TRAIN_MAX_TOKENS; option markers must not be truncated.")
        for key in ("input_ids","attention_mask"):
            b[key]=b[key][:,:TRAIN_MAX_TOKENS]
        with torch.no_grad(),torch.autocast("cuda",dtype=amp_dtype):
            teacher_logits,_=teacher(**b)
        optimizer.zero_grad(set_to_none=True)
        with torch.autocast("cuda",dtype=amp_dtype):
            student_logits,_=student(**b)
            temperature=2.
            target=F.softmax(teacher_logits.float()/temperature,dim=-1)
            predicted=F.log_softmax(student_logits.float()/temperature,dim=-1)
            loss=F.kl_div(predicted,target,reduction="batchmean")*temperature**2
        if not bool(torch.isfinite(loss)):
            raise RuntimeError(f"Nonfinite distillation loss at step {step}")
        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(trainable,1.)
        scaler.step(optimizer)
        scaler.update()
        if step%CHECK_EVERY_STEPS==0 or step==MAX_TRAIN_STEPS:
            q=quality(student,probe_items,PROBE_REF)
            passed=quality_pass(q)
            perfect=exact_perfect(q)
            confirmed=False
            if perfect and STOP_ON_PERFECT_MATCH:
                # Confirm on a second full-probe evaluation of same frozen weights.
                confirmed=exact_perfect(quality(student,probe_items,PROBE_REF))
            improved=rank_checkpoint(q)>rank_checkpoint(best_quality)
            if improved:
                best_quality=q
                best_step=step
                save_best(step,q,"perfect_verified" if confirmed else "best_fidelity_first")
            rec=dict(step=step,loss=float(loss.detach().cpu()),quality=q,
                     quality_pass=passed,perfect_top1=perfect,
                     perfect_confirmed=confirmed,selected=improved)
            training_log.append(rec)
            LOG.write_text(json.dumps(training_log,indent=2),encoding="utf-8")
            print(f"STEP {step}: agreement={q['teacher_agreement']:.4%} "
                  f"gold={q['gold_accuracy']:.4%} JS={q['mean_js']:.6f} "
                  f"selected={improved} perfect={confirmed}")
            if confirmed:
                # Even if step equals best because of ties, save this verified
                # perfect snapshot to disk before exiting.
                best_quality=q
                best_step=step
                save_best(step,q,"perfect_first_accepted")
                early_stop_reason="perfect_teacher_top1_and_gold_preserved"
                print("EARLY STOP: all 120 teacher decisions match; JS and gold gates pass.")
                break

restore_from_weights(BEST_FILE)
student.eval().requires_grad_(False)
final_probe=quality(student,probe_items,PROBE_REF)
assert quality_pass(final_probe)==quality_pass(best_quality)
selection_record=dict(version="V5.1",selected_step=best_step,
    max_train_steps=MAX_TRAIN_STEPS,check_every_steps=CHECK_EVERY_STEPS,
    early_stop_reason=early_stop_reason,best_quality=best_quality,
    final_probe=final_probe,adapter_checkpoint=str(BEST_FILE),
    selected_spec_sha256=spec_hash,steps_logged=len(training_log))
(OUT/"early_stop_summary.json").write_text(
    json.dumps(selection_record,indent=2),encoding="utf-8")
print("FINAL SELECTED STEP:",best_step,"STOP REASON:",early_stop_reason)
print("FROZEN PROBE:",final_probe)
````
